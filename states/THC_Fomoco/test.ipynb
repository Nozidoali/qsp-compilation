```python
import numpy as np
from thc_state_prep import THCFactors, build_nonorthogonal_prepare_terms

# chi[p, mu], shape = (n_orb, M)
# zeta[mu, nu], shape = (M, M)
chi = np.load("chi.npy")
zeta = np.load("zeta.npy")

factors = THCFactors(chi=chi, zeta=zeta)

# 1) Compute G_{pqrs}
G = factors.compute_G()
print(G.shape)
print(G[0, 1, 2, 3])

# 2) Compute lambda values
lam_direct = factors.lambda_thc_direct()          # Eq. (9)
lam_zeta = factors.lambda_zeta_nonorthogonal()    # Eq. (13), with chi normalized to paper convention

print("lambda_direct =", lam_direct)
print("lambda_zeta   =", lam_zeta)

# 3) Build the main PREPARE-state coefficients for Eq. (27)
# t_l are the one-body coefficients appearing in Eq. (26),
# not the raw matrix T_{pq}.
t = np.load("t.npy")

prep = build_nonorthogonal_prepare_terms(
    t=t,
    zeta=factors.normalize_columns_paper_convention().zeta
)

print("lambda_total =", prep["lambda_total"])
print("first one-body term =", prep["one_body_terms"][0])
print("first two-body term =", prep["two_body_terms"][0])
```
